# WVF quickstart

This notebook shows the three supported usage modes for the Wide View Filter code in this repository.

1. `wvf_metal` with the native CPU FFT backend.
2. `wvf_metal` with the CUDA/VkFFT backend.
3. The pure Python reference fallback from `src/wvf`.


## Imports and test image

```python
import numpy as np

image = np.random.default_rng(0).random((512, 512), dtype=np.float32)
image.shape
```

## 1. CPU FFT with `wvf_metal`

```python
from wvf_metal import backend_info, components

print(backend_info())

cpu_result = components(
    image,
    radius=15,
    degree=11,
    normalize_coords=True,
    variant="fft",
    fft_backend="cpu",
)

cpu_result.gx.shape, cpu_result.magnitude.dtype
```

## 2. GPU FFT with `wvf_metal`

This requires the CUDA/VkFFT backend to be available on Linux.

```python
gpu_result = components(
    image,
    radius=15,
    degree=11,
    normalize_coords=True,
    variant="fft",
    fft_backend="vkfft",
)

gpu_result.gx.shape, gpu_result.magnitude.dtype
```

## 3. Pure Python fallback

This cell requires the full repository checkout and `python -m pip install .` from the repository root.

```python
from wvf import wvf_radius_gradients_cpu

gx_ref, gy_ref = wvf_radius_gradients_cpu(
    image,
    radius=15,
    order=11,
    normalize_coords=True,
)

gx_ref.shape, gx_ref.dtype
```
